```python
import numpy as np
def sigmoid(x):
  return 1/(1+np.exp(-x))
values=[-2,-1,0,1,2]
for v in values:
  print(v,"=>",sigmoid(v))
```

```python
def relu(x):
  return max(0,x)
values=[-2,-1,0,1,2]
for v in values:
  print(v,"=>",relu(v))
```

```python
import numpy as np
values = np.array([[0,0],[0,1],[1,0],[1,1]])
weight=[1,1]
bias=-1.5
for v in values:
  output=np.dot(v,weight)+bias
  if output>=0:
    print(v,"=>",1)
  else:
    print(v,"=>",0)

```

```python
import torch
import torch.nn as nn
x=torch.randn(1,1,28,28)
cnn=nn.Sequential(
    nn.Conv2d(1,8,3),
    nn.ReLU(),
    nn.MaxPool2d(2)

                  )
output=cnn(x)
print("input shape:",x.shape)
print("output shape:",output.shape)
```

```python
import numpy as np
image = np.random.rand(28,28)
noise = np.random.normal(
    0,0.2,image.shape
)
noisy_image = image+noise
noisy_image = np.clip(
    1,2,noisy_image
)
print("orginal image shape :",image.shape)
print("noisy image shape: ",noisy_image.shape)
```

```python
import numpy as np
orginal_image=np.array([0.2,0.4,0.6,0.8])
reconstructed_image=np.array([0.3,0.5,0.5,0.7])
mse=np.mean((orginal_image-reconstructed_image)**2)
print("mean square error :",mse)

```
